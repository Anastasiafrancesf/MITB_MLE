```python

def plot_distribution(df, col_name, clean_underscore=True, bins=50, cap_percentile=0.99):
    series = df[col_name].astype(str)
    
    if clean_underscore:
        series = series.str.replace('_', '', regex=False).str.strip()
    
    clean = pd.to_numeric(series, errors='coerce')
    clean = clean.replace([np.inf, -np.inf], np.nan).dropna()

    print(f"--- {col_name} ---")
    print(f"Total rows: {len(df)} | Valid numeric: {len(clean)} | Dropped/NaN: {len(df) - len(clean)}")
    print(clean.describe())

    # cap x-axis at a percentile so outlier tails don't flatten the histogram
    upper = clean.quantile(cap_percentile)
    plot_data = clean[clean <= upper]

    plt.figure(figsize=(10, 5))
    plt.hist(plot_data, bins=bins, edgecolor='black')
    plt.xlabel(col_name)
    plt.ylabel('Count')
    plt.title(f'Distribution of {col_name} (capped at {int(cap_percentile*100)}th percentile)')
    plt.show()

    return clean

# just change this to whichever column you want to check
col_name = 'Changed_Credit_Limit'
cleaned_series = plot_distribution(fin, col_name)
```
